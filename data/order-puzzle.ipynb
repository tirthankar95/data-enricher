```python
import pandas as pd
df = pd.read_csv('grid-puzzles.csv')

df['Difficulty'] = df['Difficulty'].map(
    {
        'Easy': 0,
        'Moderate': 1,
        'Challenging': 2
    }
)
df = df.sort_values(by='Difficulty', kind='stable').reset_index(drop=True)
df['Difficulty'] = df['Difficulty'].map(
    {
        0: 'Easy',
        1: 'Moderate',
        2: 'Challenging'
    }
)
df = df.sort_values(by='Grid', kind='stable').reset_index(drop=True)
df.head(n=10)
```

```python
df.shape
```

```python
difficulty_levels = ['Easy', 'Moderate', 'Challenging']
grid_sizes = ['3x4', '3x5', '4x4', '4x5', '4x6', '4x7']
LIMIT = 120

for difficulty in difficulty_levels:
    for grid_size in grid_sizes:
        assert df[(df['Difficulty'] == difficulty) & (df['Grid'] == grid_size)].shape[0] == LIMIT, \
            f'{difficulty} {grid_size} has {df[(df["Difficulty"] == difficulty) & (df["Grid"] == grid_size)].shape[0]} puzzles, expected {LIMIT}'
```

```python
_df = df.drop_duplicates()
if _df.shape[0] != df.shape[0]:
    print(f'Duplicates found: {_df.shape[0]} vs {df.shape[0]}')

df.to_csv('grid-puzzles-ordered.csv', index=False)
```
